```python
import pandas as pd
df = pd.read_excel('Dataset for Data Analytics.xlsx')
df.head()
```

```python
print(df.isnull().sum())
```

```python
df['CouponCode'] = df['CouponCode'].fillna('No Coupon')
```

```python
print(df.duplicated().sum())
```

```python
df['Date'] = pd.to_datetime(df['Date']).dt.date
df['Quantity'] = pd.to_numeric(df['Quantity'])
df['UnitPrice'] = pd.to_numeric(df['UnitPrice'])
```

```python
df['CalculatedPrice'] = df['Quantity'] * df['UnitPrice']

print(df[['TotalPrice', 'CalculatedPrice']].head())
```

```python
print((df['TotalPrice'] != df['CalculatedPrice']).sum())
```

```python
print(
    (round(df['TotalPrice'], 2) != round(df['CalculatedPrice'], 2)).sum()
)
```

```python
df = df.drop(columns=['TotalPrice'])  
```

```python
print((df['Quantity'] < 0).sum())
```

```python
print((df['UnitPrice'] <= 0).sum())
```

```python
df['Product'] = df['Product'].str.strip()
df['PaymentMethod'] = df['PaymentMethod'].str.strip()
```

```python
df.to_excel('Cleaned_Dataset.xlsx', index=False)
```
